## Cleaning decisions

**Scope and rows to remove**
- Keep only detached single dwellings (Property Use Code = RESSD): 177,823 rows.
- Drop the 56 land-only rows (missing Total Living Area, Year Built, Building Type, and Air Conditioning; 0 dwelling units).
- Drop the 10 rows with 0 dwelling units that do have building data (inconsistent assessed values, including $0).
- Drop the 30 rows with 2 or fewer rooms (room count is unreliable).
- Drop the 7 rows whose Property Class 1 is not RESIDENTIAL 1 (non-standard for detached houses).
- Drop the 23 rows still missing Basement after the drops above.
- Keep the 11 houses built in 2026 and the 29 houses with 15 or more rooms; revisit them in model error analysis.
- No duplicate rows to remove: no exact duplicates, and repeated addresses appear to be different properties. The shared-location pair involves 1637 Main Street, already dropped under Property Class.
- drop the row with a 0 assessed value.
- Rooms (6 missing) and Zoning (4 missing): drop them

**Type conversions**
- Total Assessed Value (target), Total Living Area, Assessed Land Area: remove "$" and "," and convert to numbers.
- Water and Sewer Frontage Measurement: convert to numbers.
- Year Built and Rooms: convert to integers after the drops.
- Basement, Basement Finish, Fire Place, Attached Garage, Detached Garage, Air Conditioning, Pool: encode Yes/No as 1/0.

**Columns to drop**
- Leakage: Assessed Value 1 to 5 (Assessed Value 1 is identical to the target), along with Property Class and Status 2 to 5.
- Empty: Number Floors (Condo), Total Proposed Assessment Value, all Proposed fields, Unit Number, Street Suffix, Street Direction.
- Constant or nearly constant: Property Use Code, Assessment Date, Current Assessment Year, Multiple Residences, Property Class 1, Status 1, Dwelling Units.
- Identifiers (kept for reference, not used as features): Roll Number, Street Number, Street Name, Street Type, Full Address, Detail URL, Geometry, GISID.

**Feature preparation**
- Property Influences: turn the most common influences into 1/0 flags, treating missing as "none."
- Zoning: group rare codes into "Other," then encode.
- Neighbourhood Area (203 categories) and Market Region (10): encode later; keep Centroid Lat/Lon for location features.
- Water and Sewer Frontage are nearly identical: keep one or their average, and decide how to treat missing values.
- Zoning: group rare codes into Other” to “Zoning: codes with fewer than 500 houses grouped into Other

**Checks after conversion**
- keep the large-lot rows, and handle the skew in Land Area during feature preparation.
- inspect the 34 rows between 1 and 50,000, then choose a cutoff.
- Low priority: about 100 repeated Full Address values.

Property Influences contains 41 distinct labels. The most common are BUS ROUTE, LIGHT TRAFFIC, and EXTERNAL CORNER. I will create 15 binary flags for influences appearing in at least 2,000 houses, treating missing values as none.

```python
light = clean_data["Property Influences"].str.contains("LIGHT TRAFFIC", na=False)
medium = clean_data["Property Influences"].str.contains("MEDIUM TRAFFIC", na=False)
heavy = clean_data["Property Influences"].str.contains("HEAVY TRAFFIC", na=False)

traffic_count = light.astype(int) + medium.astype(int) + heavy.astype(int)

(traffic_count >= 2).sum()
```

```python
clean_data["Traffic Level"] = (
    clean_data["Property Influences"].str.contains("LIGHT TRAFFIC", na=False).astype(int) * 1 +
    clean_data["Property Influences"].str.contains("MEDIUM TRAFFIC", na=False).astype(int) * 2 +
    clean_data["Property Influences"].str.contains("HEAVY TRAFFIC", na=False).astype(int) * 3
)
```

LIGHT TRAFFIC, MEDIUM TRAFFIC, and HEAVY TRAFFIC never appear together (0 houses have more than one), so they were combined into a single Traffic Level column: 0 for no traffic influence, 1 for light, 2 for medium, and 3 for heavy. This uses one column instead of three, though it assumes the levels are evenly spaced.

```python
clean_data["Traffic Level"].value_counts().sort_index()
```

```python
influence_counts = influences.value_counts()

traffic_names = ["LIGHT TRAFFIC", "MEDIUM TRAFFIC", "HEAVY TRAFFIC"]

selected_influences = [
    influence
    for influence in influence_counts[influence_counts >= 2000].index
    if influence not in traffic_names
]
```

```python
len(selected_influences)
```

```python
influence_labels = clean_data["Property Influences"].fillna("").str.split(",")
for influence in selected_influences:
    clean_data[f"Influence: {influence}"] = influence_labels.apply(
        lambda labels: int(influence in [label.strip() for label in labels])
    )
```

```python
for influence in selected_influences:
    flag_sum = clean_data[f"Influence: {influence}"].sum()
    expected_count = influence_counts.get(influence, 0)

    status = "MATCH" if flag_sum == expected_count else "MISMATCH"

    print(
        f"{influence}: flag = {flag_sum}, "
        f"expected = {expected_count}, {status}"
    )
```

Property Influences was turned into 12 flag columns, one for each influence found on at least 2,000 houses (excluding the three traffic labels, which became Traffic Level). Each entry was split into its separate labels and matched exactly, and every flag's total equals the influence's count from the earlier split (for example, 22,217 houses for BUS ROUTE and 11,219 for PARK). Missing values were treated as "none" (all flags 0). The original text column was then dropped.
